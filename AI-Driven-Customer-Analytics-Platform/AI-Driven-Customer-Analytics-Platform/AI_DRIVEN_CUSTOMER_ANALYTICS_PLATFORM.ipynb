```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

```python
X_scaled.shape
```

```python
pd.DataFrame(X_scaled).isnull().sum()
```

```python
from sklearn.cluster import KMeans
```

```python
wcss = []

for i in range(1,11):
    kmeans = KMeans(
        n_clusters=i,
        random_state=42
    )
    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)
```

```python
plt.figure(figsize=(8,5))

plt.plot(range(1,11), wcss, marker='o')

plt.xlabel("Number of Clusters")
plt.ylabel("WCSS")
plt.title("Elbow Method")

plt.show()
```

```python
kmeans = KMeans(
    n_clusters=4,
    random_state=42
)

clusters = kmeans.fit_predict(X_scaled)
```

```python
df["Cluster"] = clusters
```

```python
%%writefile app.py

import streamlit as st
import pandas as pd
import pickle
import google.generativeai as genai

# Page Config
st.set_page_config(page_title="AI Customer Analytics", layout="wide")

st.title(" AI-Driven Customer Analytics Platform")

# Load Data
@st.cache_data
def load_data():
    return pd.read_csv("customer_segments.csv")

df = load_data()

# 1. Dataset Overview
st.header(" Dataset Overview")
st.dataframe(df.head())

# 2. Visualizations
col1, col2 = st.columns(2)

with col1:
    st.header(" Income Distribution")
    st.line_chart(df["Income"])

with col2:
    st.header(" Customer Segments")
    cluster_counts = df["Cluster"].value_counts()
    st.bar_chart(cluster_counts)

# 3. AI Strategy & Insights
st.divider()
st.header(" AI Business Strategy Report")

# Configure Gemini (Using the key provided in current session)
genai.configure(api_key="YOUR_API_KEY_HERE")
ai_model = genai.GenerativeModel("gemini-1.5-flash")

if st.button("Generate AI Business Analysis"):
    with st.spinner("Senior Analyst is reviewing data..."):
        prompt = f"""
        You are a senior business analyst.
        The customer analytics dataset has the following statistics:
        - Average Income: {df['Income'].mean():.2f}
        - Average Spending: {df['Total_Spending'].mean():.2f}
        - Number of Customers: {len(df)}
        - Number of Customer Segments: {df['Cluster'].nunique()}

        Provide a professional report including Executive Summary, Behavior Insights, and Marketing Recommendations.
        """
        response = ai_model.generate_content(prompt)
        st.markdown(response.text)

# 4. Prediction Section
st.divider()
st.header(" Predict Customer Spending")

# Load model
with open("rf_reg_model.pkl", "rb") as file:
    prediction_model = pickle.load(file)

with st.expander("Enter Customer Details", expanded=True):
    income = st.number_input("Annual Income", value=float(df['Income'].mean()))
    age = st.number_input("Age", min_value=18, max_value=100, value=40)
    children = st.number_input("Number of Children", min_value=0, max_value=10, value=0)
    recency = st.number_input("Recency (Days since last purchase)", min_value=0, value=30)
    total_purchases = st.number_input("Total Number of Purchases", min_value=0, value=5)

if st.button("Calculate Predicted Spending"):
    input_data = pd.DataFrame({
        "Income": [income],
        "Age": [age],
        "Children": [children],
        "Recency": [recency],
        "Total_Purchases": [total_purchases]
    })
    prediction = prediction_model.predict(input_data)
    st.success(f"### Estimated Total Spending: ${prediction[0]:,.2f}")
```

```python
import pandas as pd

df = pd.read_csv("customer_segments.csv")
```

```python
from scipy.stats import f_oneway

cluster0 = df[df["Cluster"]==0]["Income"]
cluster1 = df[df["Cluster"]==1]["Income"]
cluster2 = df[df["Cluster"]==2]["Income"]
cluster3 = df[df["Cluster"]==3]["Income"]

f_stat, p_value = f_oneway(
    cluster0,
    cluster1,
    cluster2,
    cluster3
)

print("F-statistic:", f_stat)
print("P-value:", p_value)
```

```python
if p_value < 0.05:
    print("There is a statistically significant difference in Income among clusters.")
else:
    print("There is no statistically significant difference in Income among clusters.")
```
